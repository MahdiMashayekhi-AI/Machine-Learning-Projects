## Detecting the outliers using 3 standard deviations rule:

```python
from scipy.stats import zscore

z_scores = zscore(df)

threshold = 3

outliers_indices = np.where(np.abs(z_scores) > threshold)

num_outliers = len(outliers_indices[0])
num_outliers
```

## IQR (Interquartile Range)

```python
column_to_analyze = 'Amount'

Q1 = df[column_to_analyze].quantile(0.25)
Q2 = df[column_to_analyze].quantile(0.50)
Q3 = df[column_to_analyze].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[(df[column_to_analyze] < lower_bound) | (df[column_to_analyze] > upper_bound)]
```

```python
outliers.shape[0]
```

```python
# Find the outliers for entire data frame
Q1 = df.quantile(0.25)
Q3 = df.quantile(0.75)

lower_bounds = Q1 - 1.5 * (Q3 - Q1)
upper_bounds = Q3 + 1.5 * (Q3 - Q1)
```

```python
((df > upper_bounds) | (df < lower_bounds)).sum()
```

```python
((df > upper_bounds) | (df < lower_bounds)).sum().sum()
```

## Handeling Outliers: 

### 1.Removal: We can remove outliers from the dataset

```python
# Example of removing outliers based on an IQR-Based method
data1 = df[(df[column_to_analyze] >= lower_bound) & (df[column_to_analyze] <= upper_bound)]
print("Rows before removing the outliers ",df.shape[0])
print(" ")
print("Rows after removing the outliers ",data1.shape[0])
print(" ")
print("Rows lost ", df.shape[0] - data1.shape[0])
```

```python
data2 = df.copy()
data2.loc[data2[column_to_analyze] < lower_bound, column_to_analyze] = Q2
data2.loc[data2[column_to_analyze] > upper_bound, column_to_analyze] = Q2

data2[(data2[column_to_analyze] < lower_bound) | (data2[column_to_analyze] > upper_bound)].shape[0]
```

### Using algorithmic approach to treat outliers

```python
data5 = df.copy()
data5[(data5 > upper_bounds) | (data5 < lower_bounds)] = np.nan
data5.isnull().sum().sum()
```

```python
from sklearn.preprocessing import RobustScaler
from sklearn.impute import KNNImputer

# Apply RobustScaler to handle outliers
scaler = RobustScaler()
scaled_data = scaler.fit_transform(data5)
scaled_df = pd.DataFrame(scaled_data, columns=data5.columns)

# Create a KNNImputer instance
knn_imputer = KNNImputer(n_neighbors=5)

# Use KNNImputer to impute missing values
imputed_data = knn_imputer.fit_transform(scaled_df)
imputed_df = pd.DataFrame(imputed_data, columns=scaled_df.columns)

# Inverse transform to get data back to the original scale
original_scale_data = scaler.inverse_transform(imputed_data)
original_scale_df = pd.DataFrame(original_scale_data, columns=scaled_df.columns)
```

```python
fig, axes = plt.subplots(1, 2)
sns.histplot(x = 'Amount', data = original_scale_df, bins = 30, ax = axes[0])
axes[0].set_title('Histogram of Amount')

sns.boxplot(x = 'Amount', data = original_scale_df, ax = axes[1])
axes[1].set_title('Boxplot of Amount')
plt.show()
```
