# Visualization using Matplotlib

## Import libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

## Histogram

```python
data = [1, 2, 2, 3, 3, 3, 4, 4, 5, 5, 6]

plt.hist(data, bins=5)
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title("Histogram")
plt.show()
```

## Box Plot

```python
np.random.seed(0)
data = np.random.randn(100)

plt.boxplot(data)
plt.xlabel("Data")
plt.title("Box Plot")
plt.show()
```

## Pie Plot

```python
value = [538, 431]
index = ['Male', 'Female']
columns = ['Passengers']

data = pd.DataFrame(data=value, index=index, columns=columns)
data
```

```python
plt.figure(figsize=(5, 5))
plt.pie(data['Passengers'], labels=data.index, autopct='%0.2f%%')
plt.title("Distribution of Passengers by Gender")
plt.show()
```

## Bar Plot

```python
x = ['Graduate', 'post Graduate', 'High School', 'Diploma']
y = [25, 42, 30, 45]

plt.figure(figsize=(7, 5))
plt.bar(x, y, color='blue')
plt.title("Count Plot")
plt.show()
```

## Scatter Plot

```python
np.random.seed(0)
x = np.random.rand(100) * 10

noise = np.random.normal(0, 1, size=len(x))
y = 2 * x + 3 + noise

plt.figure(figsize=(8, 6))
plt.scatter(x, y, color='blue', label='Data Points')
plt.xlabel('X')
plt.ylabel('Y')
plt.title("X-Y Scatter Plot")
plt.show()
```

## Box Plot for two or more categorical features

```python
np.random.seed(0)

df = pd.DataFrame({'Team_A' : np.random.randint(20, 50, 30),
                   'Team_B' : np.random.randint(40, 70, 30)})

plt.boxplot(df, labels=df.columns)
plt.xlabel("Team")
plt.ylabel("Score")
plt.show()
```

```python
df = pd.DataFrame({'Selected' : [40, 70], 'Rejected' : [10, 60]}, index=['Site A', 'Site B'])
df
```

```python
normalized_df = df.div(df.sum(axis=0), axis=1)
normalized_df
```

## Stacked Bar for categorical features

```python
categories = ['Selected', 'Rejected']
values1 = normalized_df.loc['Site A']
values2 = normalized_df.loc['Site B']

x = np.arange(len(categories))

plt.bar(x, values1, width=0.4, label='Site A')
plt.bar(x, values2, width=0.4, bottom=values1, label='Site B')

plt.xlabel("Categories")
plt.ylabel("Values")
plt.title("Stacked Bar Plot")
plt.legend()
plt.xticks(x, categories)
plt.show()
```

## Histogram with Text customization

```python
data = [1, 2, 2, 3, 3, 3, 4, 4, 5, 5, 6]

plt.hist(data, bins=5, color='blue')

plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram')

counts, bins, _ = plt.hist(data, bins=5)
print(counts)
print(bins)

for i, count in enumerate(counts):
  plt.text(bins[i] + 0.4, count + 0.1, str(int(count)), ha='left')

plt.grid(False)
plt.show()
```

## Bar Plot with Text customization

```python
x = ['Graduate', 'post Graduate', 'High School', 'Diploma']
y = [25, 42, 30, 45]

plt.figure(figsize=(7, 5))
plt.bar(x, y, color='blue')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title("Count Plot")
plt.ylim(0, 50)

for i, v in enumerate(y):
  plt.text(i, v + 1, str(v), ha='center', va='bottom')

plt.show()
```

```python
x = ['Graduate', 'post Graduate', 'High School', 'Diploma']
y = [25, 42, 30, 45]

sorted_pairs = sorted(zip(x, y), key=lambda pair: pair[1], reverse=True)
print(sorted_pairs)

x_sorted, y_sorted = zip(*sorted_pairs)
print(x_sorted)
print(y_sorted)

plt.figure(figsize=(7, 5))
plt.bar(x_sorted, y_sorted, color='blue')
plt.xlabel('Categories')
plt.ylabel('Values')
plt.title("Count Plot")
plt.ylim(0, 50)

for i, v in enumerate(y_sorted):
  plt.text(i, v + 1, str(v), ha='center', va='bottom')

plt.show()
```

## Customized Box Plot

```python
data = [1, 2, 3, 4, 5, 6, 7, 8, 9]

plt.boxplot(data, notch=True, vert=False, patch_artist=True,
            boxprops={'facecolor':'lightblue'},
            whiskerprops={'color':'blue'},
            medianprops={'color':'red', 'linewidth':2},
            labels=['Data'], widths=0.5)
plt.xlabel('Values')
plt.title('Customized Box Plot')
plt.show()
```

```python
data = [10, 15, 20, 25, 30, 35, 40, 45, 50]

plt.boxplot(data, patch_artist=True,
            boxprops={'facecolor':'skyblue', 'edgecolor':'blue'},
            whiskerprops={'color':'red'},
            medianprops={'color':'orange'},
            capprops={'color':'green'})

# Calculate the quantile
q1 = np.percentile(data, 25)
q2 = np.percentile(data, 50)
q3 = np.percentile(data, 75)

plt.text(1, q1 + 2, f'Q1: {q1}', ha='center', va='bottom', color='blue')
plt.text(1, q2 + 2, f'Q2: {q2}', ha='center', va='bottom', color='blue')
plt.text(1, q3 + 2, f'Q3: {q3}', ha='center', va='bottom', color='blue')

plt.xlabel('Boxplot', color='purple')
plt.ylabel('Values', color='purple')
plt.title('Customized Box Plot')
plt.xticks([1], ['Data'], color='purple')
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```

## Customized Pie Plot

```python
labels = ['Category A', 'Category B', 'Category C', 'Category D']
sizes = [15, 30, 45, 10]
colors = ['blue', 'green', 'red', 'purple']

plt.figure(figsize=(6, 6))
plt.pie(sizes, labels=labels, colors=colors, autopct='%1.1f%%', startangle=90)
plt.axis('equal')
plt.title('Pie Chart')
plt.show()
```

## Customized Stack Bar

```python
data = {'Selected' : [40, 70], 'Rejected' : [10, 60]}
index = ['Site A', 'Site B']
df = pd.DataFrame(data, index=index)
normalized_df = df.div(df.sum(axis=0), axis=1)

categories = ['Selected', 'Rejected']
values1 = normalized_df.loc['Site A']
values2 = normalized_df.loc['Site B']

x = np.arange(len(categories))

plt.bar(x, values1, width=0.4, label='Site A')
plt.bar(x, values2, width=0.4, bottom=values1, label='Site B')

plt.xlabel('Categories')
plt.ylabel('Values')
plt.title('Stacked Bar Plot')
plt.legend()
plt.xticks(x, categories)

for i, (v1, v2) in enumerate(zip(values1, values2)):
  plt.text(i, v1/2, f'{round(v1 * 100, 2)}%', ha='center', va='center', color='white', fontweight='bold')
  plt.text(i, v1 + v2/2, f'{round(v2 * 100, 2)}%', ha='center', va='center', color='white', fontweight='bold')

plt.show()
```

## Customized Line Plot

```python
x = [1, 2, 3, 4, 5]
y = [10, 12, 5, 8, 6]

plt.plot(x, y, label='Line-1', color='red', linestyle='--', marker='o', markersize=8)

for i, j in zip(x, y):
  plt.text(i, j + 0.5, str(j), ha='center', va='center', color='blue')

plt.ylim(0, 13)
plt.xticks(x)
plt.grid(True, alpha=0.7)
plt.xlabel('X-axis', fontsize=12, fontweight='bold', color='purple')
plt.ylabel('Y-axis', fontsize=12, fontweight='bold', color='purple')
plt.title('Line Plot', fontsize=16, fontweight='bold', color='purple')
plt.legend()
plt.show()
```

```python
x = [1, 2, 3, 4, 5]
y1 = [2, 4, 6, 8, 10]
y2 = [3, 5, 7, 9, 11]

plt.plot(x, y1, color='blue', label='Series 1')
plt.plot(x, y2, color='red', label='Series 2')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.title('Line Plot with Multiple Series')
plt.legend(loc='lower right', fontsize='x-large')
plt.grid(True, alpha=0.5)
plt.show()
```

```python
x = np.linspace(0, 10, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)
y4 = np.power(x, 2)

fig, axis = plt.subplots(2, 2, figsize=(15, 10))

axis[0, 0].plot(x, y1)
axis[0, 0].set_title('Sin(x)')
axis[0, 0].set_xlabel('X-axis Label 1')
axis[0, 0].set_ylabel('Y-axis Label 1')

axis[0, 1].plot(x, y2)
axis[0, 1].set_title('Cos(x)')
axis[0, 1].set_xlabel('X-axis Label 2')
axis[0, 1].set_ylabel('Y-axis Label 2')


axis[1, 0].plot(x, y3)
axis[1, 0].set_title('Tan(x)')
axis[1, 0].set_xlabel('X-axis Label 3')
axis[1, 0].set_ylabel('Y-axis Label 3')

axis[1, 1].plot(x, y4)
axis[1, 1].set_title('x^2')
axis[1, 1].set_xlabel('X-axis Label 4')
axis[1, 1].set_ylabel('Y-axis Label 4')

plt.tight_layout()
plt.show()
```

```python
x = np.linspace(-10, 10, 200)

y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)

fig, axis = plt.subplots(2, 2, figsize=(15, 10))

axis[0, 0].plot(x, y1, 'b')
axis[0, 0].set_title('Sin(x)')
axis[0, 0].set_xlabel('X-axis')

axis[0, 1].plot(x, y2, 'r')
axis[0, 1].set_title('Cos(x)')
axis[0, 1].set_xlabel('X-axis')

axis[1, 0].plot(x, y3, 'g')
axis[1, 0].set_title('Tan(x)')
axis[1, 0].set_xlabel('X-axis')

fig.delaxes(axis[1, 1])

plt.suptitle('Grid of Subplots (Create with plt.subplots)')
plt.tight_layout()
plt.show()
```

```python
x = np.linspace(0, 2 * np.pi, 100)
y1 = np.sin(x)
y2 = np.cos(x)
y3 = np.tan(x)

plt.figure(figsize=(15, 4))

plt.subplot(1, 3, 1)
plt.plot(x, y1, 'g')
plt.title('Sin(x)')
plt.xlabel('X-axis')

plt.subplot(1, 3, 2)
plt.plot(x, y2, 'r')
plt.title('Cos(x)')
plt.xlabel('X-axis')

plt.subplot(1, 3, 3)
plt.plot(x, y3, 'b')
plt.title('Tan(x)')
plt.xlabel('X-axis')

plt.tight_layout()
plt.show()
```

```python
plt.rcParams['figure.figsize'] = (8, 4)
plt.rcParams['figure.dpi'] = 100
plt.rcParams['font.size'] = 12
plt.rcParams['font.weight'] = 'bold'
plt.rcParams['font.family'] = 'serif'
plt.rcParams['axes.labelsize'] = 14
plt.rcParams['axes.labelweight'] = 'bold'
plt.rcParams['axes.titlesize'] = 16

x = [1, 2, 3, 4, 5]
y = [10, 12, 5, 8, 6]

plt.plot(x, y, label='Line-1', color='red', linestyle='--', marker='o', markersize=8)

plt.title('Line Plot')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()
plt.grid(True, alpha=0.)
plt.show()
```

```python
x = np.linspace(0, 2 * np.pi, 100)
y = np.sin(x)

peak_index = np.argmax(y)
x_peak = x[peak_index]
y_peak = y[peak_index]

plt.plot(x, y, label='Sine Wave')
plt.scatter(x_peak, y_peak, color='red', label='Peak Point')

plt.annotate('Peak', xy=(x_peak, y_peak), xytext=(x_peak + 0.1, y_peak - 0.5),
             arrowprops=dict(facecolor='blue', arrowstyle='->'))

plt.title('Sine Wave with Peak Point')
plt.xlabel('X-axis')
plt.ylabel('Y-axis')
plt.legend()
plt.grid(True, alpha=0.5)
plt.show()
```
